```python
blank = " "
x_val = "X"
o_val = "O"

player_1_score = 0
player_2_score = 0

game_over = False

all_games_over = False

current_player = "Player 1"

matrix = [
    [blank,blank,blank],
    [blank,blank,blank],
    [blank,blank,blank]
]

def printing_grid():
    print(printing_row(0))
    print("_   _   _")
    print(printing_row(1))
    print("_   _   _")
    print(printing_row(2))

def printing_row(i):
    return matrix[i][0] + " | " + matrix[i][1] + " | " + matrix[i][2]

def checking_if_row_matches(i):
    global player_1_score, player_2_score
    
    if (matrix[i][0] == x_val) and (matrix[i][1] == x_val) and (matrix[i][2] == x_val):
        player_1_score += 1
        print(f"Player 1 Wins through Row {i+1}")
        return True
    elif (matrix[i][0] == o_val) and (matrix[i][1] == o_val) and (matrix[i][2] == o_val):
        player_2_score += 1
        print(f"Player 2 Wins through Row {i+1}")
        return True
    else:
        return False

def checking_if_column_matches(i):
    global player_1_score, player_2_score
    
    if (matrix[0][i] == x_val) and (matrix[1][i] == x_val) and (matrix[2][i] == x_val):
        player_1_score += 1
        print(f"Player 1 Wins through Column {i+1}")
        return True
    elif (matrix[0][i] == o_val) and (matrix[1][i] == o_val) and (matrix[2][i] == o_val):
        player_2_score += 1
        print(f"Player 2 Wins through Column {i+1}")
        return True
    else:
        return False

def checking_if_diagonal_matches():
    global player_1_score, player_2_score
    
    if (matrix[0][0] == x_val) and (matrix[1][1] == x_val) and (matrix[2][2] == x_val):
        player_1_score += 1
        print("Player 1 Wins Diagonally")
        return True
    elif (matrix[0][0] == o_val) and (matrix[1][1] == o_val) and (matrix[2][2] == o_val):
        player_2_score += 1
        print("Player 2 Wins Diagonally")
        return True
    elif (matrix[0][2] == x_val) and (matrix[1][1] == x_val) and (matrix[2][0] == x_val):
        player_1_score += 1
        print("Player 1 Wins Diagonally")
        return True
    elif (matrix[0][2] == o_val) and (matrix[1][1] == o_val) and (matrix[2][0] == o_val):
        player_2_score += 1
        print("Player 2 Wins Diagonally")
        return True
    else:
        return False

def checking_if_game_over():
    for i in range(3):
        if checking_if_row_matches(i):
            return True

    for i in range(3):
        if checking_if_column_matches(i):
            return True

    if checking_if_diagonal_matches():
        return True

    return False

def playing_the_game():
    global game_over, current_player
    
    while not game_over:
        try:
            print("________")
            print("")
            print(current_player)
            print("________")
            print("")
            
            row_num = int(input("Enter row number (1-3): "))
            column_num = int(input("Enter column number (1-3): "))
            
            if matrix[row_num-1][column_num-1] == blank:
                if current_player == "Player 1":
                    matrix[row_num-1][column_num-1] = x_val
                    current_player = "Player 2"
                else:
                    matrix[row_num-1][column_num-1] = o_val
                    current_player = "Player 1"
    
                print("")
                printing_grid()
                print("")
    
                all_done = True
                for i in range(0,3):
                    for j in range(0,3):
                        if matrix[i][j] != blank:
                            all_done = False
                if all_done:
                    print("Game's gone")
                    game_over = True
                    break
        
                if checking_if_game_over():
                    game_over = True
                
        
            else:
                print("Error, a piece is already there you bum.")
    
        except ValueError:
            print("")
            print("Value error or sumn")
        except IndexError:
            print("")
            print("Index error or sumn (make your numbers between 1-3)")

playing_the_game()

while not all_games_over:
    print("")
    symbol = str(input("Do you want to keep playing? [y/n] "))
    if symbol == 'y':
        playing_the_game()
    else:
        print("")
        print(f"Player 1 Score: {player_1_score}")
        print(f"Player 2 Score: {player_2_score}")
        print("")

        if player_1_score > player_2_score:
            print("Player 1 is the GRAND WINNER")
        elif player_2_score > player_1_score:
            print("Player 2 is the GRAND WINNER")
        else:
            print("It's a tie")  
        game_over = True
        break
```
